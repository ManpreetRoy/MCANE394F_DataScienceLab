# Question 1 – Intersection, Union and Symmetric Difference

## Problem Statement

Given two lists, find their **intersection, union, and symmetric difference**
without using Python's built-in set operations.

### Given

Two lists:

- List A
- List B

### Requirements

1. Find the intersection of the two lists.
2. Find the union of the two lists.
3. Find the symmetric difference of the two lists.
4. Do not use Python's built-in set operations.

---

## Concepts Used

- Python Lists
- Membership operator
- Loops
- Conditional statements

---

## Approach

We will implement the required operations manually using Python
lists and loops instead of using `set()`, `.union()`,
`.intersection()`, or `.symmetric_difference()`.

## Explanation

### 1. Intersection

The intersection contains elements that are present in both
List A and List B.

For the given lists:

A = [1, 2, 3, 4, 5]
B = [4, 5, 6, 7, 8]

Therefore:

Intersection = [4, 5]

We check each element of A and use the `in` operator to determine
whether the element is also present in B.

---

### 2. Union

The union contains all unique elements from both lists.

We first add the elements of A and then add the elements of B
only if they are not already present in the union list.

Therefore:

Union = [1, 2, 3, 4, 5, 6, 7, 8]

---

### 3. Symmetric Difference

The symmetric difference contains elements that belong to either
of the two lists but not to both.

Elements only in A:

[1, 2, 3]

Elements only in B:

[6, 7, 8]

Therefore:

Symmetric Difference = [1, 2, 3, 6, 7, 8]

---

## Important Restriction

Python's built-in set operations were not used.

The solution uses:

- Lists
- `for` loops
- `if` conditions
- `in` and `not in`
- `append()`

```python
A = [1,2,3,4,5]
B = [4,5,6,7,8]

# 1. Insertion
intersection = []

for item in A:
  if item in B:
    intersection.append(item)

# 2. Union

union = []

for item in A:
  if item not in union:
    union.append(item)

# 3. Symmetric Difference

Symmetric_diff = []

for item in A:
  if item not in B:
    Symmetric_diff.append(item)

for item in B:
  if item not in A:
    Symmetric_diff.append(item)


# Result

print("List A : ",A)
print("List B : ",B)
print("-------------------------")
print("Insertion : ",intersection)
print("Union : ",union)
print("Symmetric Difference : ",Symmetric_diff)
```
