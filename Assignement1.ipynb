# Question 1 – Intersection, Union and Symmetric Difference

## Problem Statement

Given two lists, find their **intersection, union, and symmetric difference**
without using Python's built-in set operations.

### Given

Two lists:

- List A
- List B

### Requirements

1. Find the intersection of the two lists.
2. Find the union of the two lists.
3. Find the symmetric difference of the two lists.
4. Do not use Python's built-in set operations.

---

## Concepts Used

- Python Lists
- Membership operator
- Loops
- Conditional statements

---

## Approach

We will implement the required operations manually using Python
lists and loops instead of using `set()`, `.union()`,
`.intersection()`, or `.symmetric_difference()`.

## Explanation

### 1. Intersection

The intersection contains elements that are present in both
List A and List B.

For the given lists:

A = [1, 2, 3, 4, 5]
B = [4, 5, 6, 7, 8]

Therefore:

Intersection = [4, 5]

We check each element of A and use the `in` operator to determine
whether the element is also present in B.

---

### 2. Union

The union contains all unique elements from both lists.

We first add the elements of A and then add the elements of B
only if they are not already present in the union list.

Therefore:

Union = [1, 2, 3, 4, 5, 6, 7, 8]

---

### 3. Symmetric Difference

The symmetric difference contains elements that belong to either
of the two lists but not to both.

Elements only in A:

[1, 2, 3]

Elements only in B:

[6, 7, 8]

Therefore:

Symmetric Difference = [1, 2, 3, 6, 7, 8]

---

## Important Restriction

Python's built-in set operations were not used.

The solution uses:

- Lists
- `for` loops
- `if` conditions
- `in` and `not in`
- `append()`

````python
A = [1,2,3,4,5]
B = [4,5,6,7,8]

# 1. Insertion
intersection = []

for item in A:
  if item in B:
    intersection.append(item)

# 2. Union

union = []

for item in A:
  if item not in union:
    union.append(item)

# 3. Symmetric Difference

Symmetric_diff = []

for item in A:
  if item not in B:
    Symmetric_diff.append(item)

for item in B:
  if item not in A:
    Symmetric_diff.append(item)


# Result

print("List A : ",A)
print("List B : ",B)
print("-------------------------")
print("Insertion : ",intersection)
print("Union : ",union)
print("Symmetric Difference : ",Symmetric_diff)
````

# Question 2 – Remove Duplicate Elements

## Problem Statement

Write a Python program to remove duplicate elements from a list
while preserving the original order of the elements.

### Example

Given:

`[10, 20, 10, 30, 20, 40, 30]`

Expected output:

`[10, 20, 30, 40]`

## Objective

To remove duplicate elements from a list without changing the
original order of the elements.

## Concept Used

- Python Lists
- `for` loop
- `if` condition
- Membership operator (`in`)
- `append()` method

## Approach

1. Create an empty list to store unique elements.
2. Traverse the original list one element at a time.
3. Check whether the element already exists in the new list.
4. If it does not exist, add it to the new list.
5. Display the resulting list.

## Explanation

The original list contains several repeated elements.

We create an empty list called `unique_numbers` to store only
the first occurrence of each element.

The `for` loop checks every element of the original list.

The condition:

`number not in unique_numbers`

checks whether the current element has already been added.

If the element is not present, `append()` adds it to the
`unique_numbers` list.

Because elements are processed from left to right, their original
order is preserved.

### Example

Original:

`[10, 20, 10, 30, 20, 40, 30, 50, 40]`

Unique elements:

`[10, 20, 30, 40, 50]`

Notice that the first occurrence of every element is retained.

## Complexity

For a list of `n` elements, the membership check can take `O(n)`
time in the worst case. Therefore, the overall time complexity
of this approach is approximately `O(n²)`.

The additional list used for storing unique elements requires
`O(n)` space.

````python
number = [10,20,10,30,20,40,30,50,40]

unique_num = []

for value in number:
  if value not in unique_num:
    unique_num.append(value)

print("Original List : ",number)
print("---------------------------")
print("Unique_number : ",unique_num)
````

## Approach

For every element, we have two choices:

1. Include the element in the current subset.
2. Exclude the element from the current subset.

We recursively process the remaining elements for both choices.

When all elements have been processed, the current subset is
added to the result.

## Explanation

The function `generate_subsets()` recursively generates all
possible subsets.

### Function Parameters

`numbers`

The original list from which subsets are generated.

`index`

The position of the current element being processed.

`current`

The subset currently being constructed.

`result`

The list containing all generated subsets.

````python
def genrate_subsets(numbers,index,current,result):

  #Base condition
  if index == len(numbers):
    result.append(current.copy())
    return

  # Choice 1 : Include the current element
  current.append(numbers[index])
  genrate_subsets(numbers,index + 1, current,result)

  # Backtracking
  current.pop()

  # choic 2 : Exclude the current element
  genrate_subsets(numbers,index + 1, current,result)

# Input List
numbers = [1,2,3]

# Store all subsets
result = []

# Genrate subsets
genrate_subsets(numbers, 0 , [],result)

#Display result
print("Original List : ", numbers)
print("All possible subsets : ")

for subset in result:
  print(subset)

print("\n Total number of subsets : ",len(result))
````

The program first opens the text file in read mode and reads its

The text is converted to lowercase so that words such as `Python`

The `split()` method divides the text into individual words.

A dictionary named `word_count` is used to store each word and

For every word:

- If the word already exists in the dictionary, its count is

- Otherwise, the word is added with a count of 1.

find the word having the highest frequency.

### Data Structure Used

Dictionary:

`word → frequency`

Example:

```text
python → 4
is → 3
easy → 1
learn → 1

````python
import string

# Read the text file
with open("sample.txt", "r") as file:
    text = file.read()

# Convert text to lowercase
text = text.lower()

# Split text into words
words = text.split()

# Remove punctuation
clean_words = []

for word in words:
    word = word.strip(string.punctuation)

    if word:
        clean_words.append(word)

# Count word frequency
word_count = {}

for word in clean_words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Find the most frequent word
most_frequent_word = max(word_count, key=word_count.get)
frequency = word_count[most_frequent_word]

# Display results
print("Word Frequencies:")
for word, count in word_count.items():
    print(word, ":", count)

print("\nMost frequent word:", most_frequent_word)
print("Frequency:", frequency)
````

# Question 5 – Convert and Validate a Roman Numeral

## Problem Statement

Write a Python program to convert a Roman numeral into its
corresponding integer value and validate whether the given
Roman numeral is valid.

## Objective

To understand string processing, dictionaries, conditional
statements, and the rules used to convert Roman numerals into
integer values.

## Roman Numeral Values

| Symbol | Value |
|---|---:|
| I | 1 |
| V | 5 |
| X | 10 |
| L | 50 |
| C | 100 |
| D | 500 |
| M | 1000 |

## Common Subtractive Combinations

- IV = 4
- IX = 9
- XL = 40
- XC = 90
- CD = 400
- CM = 900

## Approach

1. Store Roman numeral symbols and their values in a dictionary.
2. Validate the input characters.
3. Convert the Roman numeral into an integer by comparing
   adjacent symbols.
4. Validate the Roman numeral using standard Roman numeral rules.
5. Display the converted value and validation result.

````python
def roman_to_integer(roman):
    values = {
        'I': 1,
        'V': 5,
        'X': 10,
        'L': 50,
        'C': 100,
        'D': 500,
        'M': 1000
    }

    total = 0

    for i in range(len(roman)):
        current = values[roman[i]]

        if i + 1 < len(roman):
            next_value = values[roman[i + 1]]

            if current < next_value:
                total -= current
            else:
                total += current
        else:
            total += current

    return total


def integer_to_roman(number):
    values = [
        (1000, 'M'),
        (900, 'CM'),
        (500, 'D'),
        (400, 'CD'),
        (100, 'C'),
        (90, 'XC'),
        (50, 'L'),
        (40, 'XL'),
        (10, 'X'),
        (9, 'IX'),
        (5, 'V'),
        (4, 'IV'),
        (1, 'I')
    ]

    result = ""

    for value, symbol in values:
        while number >= value:
            result += symbol
            number -= value

    return result


def is_valid_roman(roman):
    if not roman:
        return False

    roman = roman.upper()

    valid_symbols = set("IVXLCDM")

    for char in roman:
        if char not in valid_symbols:
            return False

    value = roman_to_integer(roman)

    if value < 1 or value > 3999:
        return False

    return integer_to_roman(value) == roman


# User input
roman = input("Enter a Roman numeral: ").strip().upper()

if is_valid_roman(roman):
    value = roman_to_integer(roman)

    print("Roman numeral:", roman)
    print("Integer value:", value)
    print("Status: Valid Roman numeral")
else:
    print("Roman numeral:", roman)
    print("Status: Invalid Roman numeral")
````
